```python
import os

base_dir = r"C:\Users\HP\OneDrive\Desktop\MediMind\archive (3)\chest_xray"

folders = [
    os.path.join(base_dir, "train"),
    os.path.join(base_dir, "test"),
    os.path.join(base_dir, "val"),
    os.path.join(base_dir, "chest_xray", "train"),
    os.path.join(base_dir, "chest_xray", "test"),
    os.path.join(base_dir, "chest_xray", "val")
]

for folder in folders:

    print("\nFolder:", folder)

    for category in ["NORMAL", "PNEUMONIA"]:

        path = os.path.join(folder, category)

        if os.path.exists(path):

            images = [
                f for f in os.listdir(path)
                if f.lower().endswith((".jpg", ".jpeg", ".png"))
            ]

            print(category, ":", len(images))
```

```python
import hashlib

def get_hashes(folder):

    hashes = set()

    for root, dirs, files in os.walk(folder):

        for file in files:

            if file.lower().endswith((".jpg", ".jpeg", ".png")):

                path = os.path.join(root, file)

                with open(path, "rb") as f:
                    file_hash = hashlib.sha256(f.read()).hexdigest()

                hashes.add(file_hash)

    return hashes

original = get_hashes(os.path.join(base_dir, "train"))
nested = get_hashes(os.path.join(base_dir, "chest_xray", "train"))

print("Original unique images:", len(original))
print("Nested unique images:", len(nested))

print("Common images:", len(original.intersection(nested)))
print("Only in original:", len(original - nested))
print("Only in nested:", len(nested - original))
```

```python
import os
import shutil
import stat

duplicate_path = os.path.join(base_dir, "chest_xray")

def remove_readonly(func, path, exc_info):
    os.chmod(path, stat.S_IWRITE)
    func(path)

if os.path.exists(duplicate_path):
    shutil.rmtree(duplicate_path, onerror=remove_readonly)
    print("Duplicate folder removed successfully!")
else:
    print("Duplicate folder does not exist.")
```
