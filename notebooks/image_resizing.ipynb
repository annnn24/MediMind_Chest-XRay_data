```python
import os
from PIL import Image
from tqdm import tqdm

input_dir = r"C:\Users\HP\OneDrive\Desktop\MediMind\archive (3)\chest_xray"

output_dir = r"C:\Users\HP\OneDrive\Desktop\MediMind\chest_xray_resized"

size = (224, 224)
```

```python
for split in ["train", "test", "val"]:

    for category in ["NORMAL", "PNEUMONIA"]:

        src_folder = os.path.join(input_dir, split, category)
        dst_folder = os.path.join(output_dir, split, category)

        os.makedirs(dst_folder, exist_ok=True)

        files = [
            f for f in os.listdir(src_folder)
            if f.lower().endswith((".jpg", ".jpeg", ".png"))
        ]

        for filename in tqdm(files, desc=f"{split}/{category}"):

            src_path = os.path.join(src_folder, filename)
            dst_path = os.path.join(dst_folder, filename)

            try:
                with Image.open(src_path) as img:
                    img = img.convert("RGB")
                    img = img.resize(size, Image.Resampling.LANCZOS)
                    img.save(dst_path)

            except Exception as e:
                print("Error processing:", filename, e)

print("All images processed!")
```
