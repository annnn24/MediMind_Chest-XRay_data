```python
import cv2
import numpy as np

# ImageNet normalization parameters
MEAN = np.array([0.485, 0.456, 0.406], dtype=np.float32)
STD = np.array([0.229, 0.224, 0.225], dtype=np.float32)


def normalize_image(image_path):
    # Read CLAHE-enhanced image
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if image is None:
        raise ValueError(f"Unable to read image: {image_path}")

    # Convert grayscale to 3 channels
    image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)

    # Scale pixel values from 0–255 to 0–1
    image = image.astype(np.float32) / 255.0

    # Apply ImageNet normalization
    normalized = (image - MEAN) / STD

    return normalized


# Test with one image
sample_path = (
    r"C:\Users\HP\OneDrive\Desktop\MediMind"
    r"\chest_xray_clahe\train\NORMAL"
)

import os

sample_file = os.listdir(sample_path)[0]
image_path = os.path.join(sample_path, sample_file)

normalized_image = normalize_image(image_path)

print("Image shape:", normalized_image.shape)
print("Data type:", normalized_image.dtype)
print("Minimum value:", normalized_image.min())
print("Maximum value:", normalized_image.max())
print("Normalization successful!")

```
